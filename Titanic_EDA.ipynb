```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("All libraries imported successfully!")
```

```python
import seaborn as sns

df = sns.load_dataset('titanic')

df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
print("Rows and Columns:", df.shape)
```

```python
sns.countplot(x='survived', data=df)
plt.title("Survival Count")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['age'].dropna(), bins=20)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
```

```python
sns.countplot(x='sex', hue='survived', data=df)
plt.title("Survival by Gender")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

```python
sns.countplot(x='pclass', hue='survived', data=df)
plt.title("Survival by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Count")
plt.show()
```

```python
sns.countplot(x='embarked', hue='survived', data=df)
plt.title("Survival by Embarked")
plt.xlabel("Embarked")
plt.ylabel("Count")
plt.show()
```

```python
sns.heatmap(df.select_dtypes(include='number').corr(), annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

```python
print("TITANIC DATASET INSIGHTS")
print("1. Total passengers:", len(df))
print("2. Total survived:", df['survived'].sum())
print("3. Total not survived:", (df['survived'] == 0).sum())
print("4. Average age:", df['age'].mean())
print("5. Missing values:", df.isnull().sum().sum())
```
